### Solution 1: Matrix Transpose

````python
# Approach 1: Using nested loops
def transpose_matrix_v1(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    
    # Create new matrix with swapped dimensions
    result = []
    for col in range(cols):
        new_row = []
        for row in range(rows):
            new_row.append(matrix[row][col])
        result.append(new_row)
    
    return result

# Approach 2: Using list comprehension (Pythonic)
def transpose_matrix_v2(matrix):
    rows = len(matrix)
    cols = len(matrix[0])
    return [[matrix[row][col] for row in range(rows)] for col in range(cols)]

# Approach 3: Using zip (Most Pythonic)
def transpose_matrix_v3(matrix):
    return [list(row) for row in zip(*matrix)]

# Test all approaches
matrix = [[1, 2, 3], [4, 5, 6]]
print("Original:", matrix)
print("Approach 1 (loops):", transpose_matrix_v1(matrix))
print("Approach 2 (comprehension):", transpose_matrix_v2(matrix))
print("Approach 3 (zip):", transpose_matrix_v3(matrix))
````

### Solution 2: Spiral Matrix Traversal

````python
# Approach: Using boundary pointers
def spiral_order_solution(matrix):
    if not matrix or not matrix[0]:
        return []
    
    result = []
    top, bottom = 0, len(matrix) - 1
    left, right = 0, len(matrix[0]) - 1
    
    while top <= bottom and left <= right:
        # Move right across top row
        for col in range(left, right + 1):
            result.append(matrix[top][col])
        top += 1
        
        # Move down right column
        for row in range(top, bottom + 1):
            result.append(matrix[row][right])
        right -= 1
        
        # Move left across bottom row (if still valid)
        if top <= bottom:
            for col in range(right, left - 1, -1):
                result.append(matrix[bottom][col])
            bottom -= 1
        
        # Move up left column (if still valid)
        if left <= right:
            for row in range(bottom, top - 1, -1):
                result.append(matrix[row][left])
            left += 1
    
    return result

# Test
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print("Matrix:")
for row in matrix:
    print(row)
print("Spiral order:", spiral_order_solution(matrix))
````

### Solution 3: Rotate Matrix 90 Degrees

````python
# Approach 1: Transpose + Reverse rows
def rotate_90_clockwise_v1(matrix):
    n = len(matrix)
    
    # Step 1: Transpose the matrix
    for i in range(n):
        for j in range(i + 1, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
    
    # Step 2: Reverse each row
    for i in range(n):
        matrix[i].reverse()

# Approach 2: Layer-by-layer rotation
def rotate_90_clockwise_v2(matrix):
    n = len(matrix)
    
    # Process each layer (outer to inner)
    for layer in range(n // 2):
        first = layer
        last = n - 1 - layer
        
        for i in range(first, last):
            offset = i - first
            
            # Save top
            top = matrix[first][i]
            
            # Left → Top
            matrix[first][i] = matrix[last - offset][first]
            
            # Bottom → Left
            matrix[last - offset][first] = matrix[last][last - offset]
            
            # Right → Bottom
            matrix[last][last - offset] = matrix[i][last]
            
            # Top → Right
            matrix[i][last] = top

# Test both approaches
import copy

original = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

matrix1 = copy.deepcopy(original)
print("Approach 1: Transpose + Reverse")
print("Before:", matrix1)
rotate_90_clockwise_v1(matrix1)
print("After:", matrix1)
print()

matrix2 = copy.deepcopy(original)
print("Approach 2: Layer-by-layer")
print("Before:", matrix2)
rotate_90_clockwise_v2(matrix2)
print("After:", matrix2)
````

## Key Takeaways

### Matrix Transpose
- Swap rows and columns: `result[j][i] = matrix[i][j]`
- Pythonic way: Use `zip(*matrix)` to unpack and zip rows
- New dimensions: m×n matrix becomes n×m matrix

### Spiral Traversal
- Use four boundary pointers: `top`, `bottom`, `left`, `right`
- Process in order: right → down → left → up
- Update boundaries after each direction
- Check boundaries before processing left/up directions

### Matrix Rotation
- **Transpose + Reverse**: Simple two-step approach
  - Transpose: swap across diagonal
  - Reverse: reverse each row
- **Layer-by-layer**: More complex but single pass
  - Rotate 4 elements at a time
  - Process outer layers first, move inward
- Both approaches modify in-place (O(1) extra space)

### General Matrix Tips
- Always check for empty matrices: `if not matrix or not matrix[0]`
- Get dimensions: `rows = len(matrix)`, `cols = len(matrix[0])`
- List comprehension for cleaner code: `[[expr for col in range(cols)] for row in range(rows)]`
- Use `zip(*matrix)` for transposing
- In-place modifications save space but require careful index management
