## Problem 2: Matrix Multiplication

### Problem Statement

Write a function `multiply_matrices(A, B)` that multiplies two matrices A and B and returns the result.

For matrix multiplication to be valid:
- Number of columns in A must equal number of rows in B
- If A is (m × n) and B is (n × p), result will be (m × p)

The element at position `[i][j]` in the result is calculated by:
- Take row i from matrix A
- Take column j from matrix B
- Multiply corresponding elements and sum them up

Formula: `result[i][j] = sum(A[i][k] * B[k][j] for all k)`

### Input Format
- `A`: First matrix (m × n)
- `B`: Second matrix (n × p)

### Output Format
- Return the product matrix (m × p)
- Return `None` if matrices cannot be multiplied

### Examples

```python

    [1, 2, 3],
    [4, 5, 6]
]

    [9, 10],
    [11, 12]
]

# Output: [[58, 64], [139, 154]]
# Calculation:
# result[0][0] = 1×7 + 2×9 + 3×11 = 7 + 18 + 33 = 58
# result[0][1] = 1×8 + 2×10 + 3×12 = 8 + 20 + 36 = 64
# result[1][0] = 4×7 + 5×9 + 6×11 = 28 + 45 + 66 = 139
# result[1][1] = 4×8 + 5×10 + 6×12 = 32 + 50 + 72 = 154

# Example 2: 2×2 multiplied by 2×2
A = [
    [1, 2],
    [3, 4]
]
B = [
    [2, 0],
    [1, 2]

# Output: [[4, 4], [10, 8]]

# Example 3: Incompatible dimensions
```
