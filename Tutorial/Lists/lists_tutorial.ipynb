## 1. What is a List?

A **list** is an ordered, mutable collection of items enclosed in square brackets `[]`.

- **Ordered** - items have a fixed position (index)
- **Mutable** - you can change, add, or remove items after creation
- **Mixed types** - a list can hold integers, floats, strings, even other lists
- **Duplicates allowed** - the same value can appear more than once

````python
# Creating lists
empty      = []                              # empty list
numbers    = [1, 2, 3, 4, 5]                # list of ints
fruits     = ["apple", "banana", "cherry"]  # list of strings
mixed      = [42, 3.14, "hello", True]      # mixed types
with_dupes = [1, 2, 2, 3, 3, 3]            # duplicates allowed

print(numbers)
print(fruits)
print(mixed)
print(f"Length of numbers: {len(numbers)}")
````

````python
# Creating lists programmatically
from_range = list(range(1, 6))   # [1, 2, 3, 4, 5]
from_str   = list("Python")      # ['P', 'y', 't', 'h', 'o', 'n']
repeated   = [0] * 5             # [0, 0, 0, 0, 0]

print(from_range)
print(from_str)
print(repeated)
````

## 2. Indexing and Slicing

### Indexing - access one element

```
  List:   [ "a",  "b",  "c",  "d",  "e" ]
Positive:     0     1     2     3     4
Negative:    -5    -4    -3    -2    -1
```

- Positive indices count **left to right** starting from `0`
- Negative indices count **right to left** starting from `-1`

````python
lst = ["a", "b", "c", "d", "e"]

print(lst[0])    # "a"  - first element
print(lst[2])    # "c"  - third element
print(lst[-1])   # "e"  - last element
print(lst[-2])   # "d"  - second-to-last
````

````python
# Lists are mutable - elements can be changed by index
lst[1] = "B"
print(lst)   # ['a', 'B', 'c', 'd', 'e']
````

### Slicing - access a sub-list

```python
lst[start : stop : step]
```
- `start` - inclusive, default `0`
- `stop`  - exclusive, default `len(lst)`
- `step`  - stride, default `1`

Slicing always returns a **new list** - the original is unchanged.

````python
lst = [10, 20, 30, 40, 50, 60, 70]

print(lst[1:4])     # [20, 30, 40]      - index 1 up to (not including) 4
print(lst[:3])      # [10, 20, 30]      - from start to index 3
print(lst[4:])      # [50, 60, 70]      - from index 4 to end
print(lst[::2])     # [10, 30, 50, 70]  - every 2nd element
print(lst[::-1])    # [70, 60, ..., 10] - reversed
print(lst[-3:])     # [50, 60, 70]      - last three elements
````

## 3. Common List Methods

| Method | What it does |
|--------|--------------|
| `append(x)` | Add `x` to the end |
| `insert(i, x)` | Insert `x` at index `i` |
| `extend(iterable)` | Add all items from iterable to the end |
| `remove(x)` | Remove the **first** occurrence of `x` |
| `pop(i)` | Remove and return item at index `i` (default: last) |
| `index(x)` | Return index of first occurrence of `x` |
| `count(x)` | Count occurrences of `x` |
| `sort()` | Sort in place (modifies original) |
| `reverse()` | Reverse in place |
| `copy()` | Return a shallow copy |

````python
lst = [3, 1, 4, 1, 5]

lst.append(9)          # add to end
print("append:", lst)

lst.insert(0, 99)      # insert 99 at index 0
print("insert:", lst)

lst.extend([7, 8])     # add multiple items
print("extend:", lst)
````

````python
lst = [3, 1, 4, 1, 5, 9, 2]

lst.remove(1)          # removes FIRST occurrence of 1
print("remove:", lst)

popped = lst.pop()     # removes & returns last element
print(f"popped: {popped} | list: {lst}")

popped2 = lst.pop(1)   # removes & returns element at index 1
print(f"popped2: {popped2} | list: {lst}")
````

````python
lst = [3, 1, 4, 1, 5, 9, 2, 6]

print("index of 5:", lst.index(5))    # 4
print("count of 1:", lst.count(1))    # 2

lst.sort()                             # sort ascending, in place
print("sorted:", lst)

lst.sort(reverse=True)                 # sort descending
print("descending:", lst)

lst.reverse()                          # reverse in place
print("reversed:", lst)
````

````python
# copy() vs direct assignment - critical distinction!
original = [1, 2, 3]

alias = original          # NOT a copy - same object in memory
alias.append(4)
print("original after alias change:", original)   # [1, 2, 3, 4] - changed!

original2 = [1, 2, 3]
copy = original2.copy()   # independent copy
copy.append(99)
print("original2 after copy change:", original2)  # [1, 2, 3] - unchanged
````

## 4. Iterating Over Lists

There are several idiomatic ways to loop through a list.

````python
fruits = ["apple", "banana", "cherry"]

# Simple iteration - when you only need values
for fruit in fruits:
    print(fruit)
````

````python
# enumerate() - when you need both index and value
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")

print()
# Start counting from 1
for i, fruit in enumerate(fruits, start=1):
    print(f"{i}. {fruit}")
````

````python
# zip() - iterate over two lists in parallel
names  = ["Alice", "Bob", "Carol"]
scores = [85, 92, 78]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

````python
# Index-based iteration - when you need to modify elements in place
numbers = [1, 2, 3, 4, 5]
for i in range(len(numbers)):
    numbers[i] *= 2
print(numbers)   # [2, 4, 6, 8, 10]
````

## 5. List Comprehensions

A concise way to build a new list from an iterable in a single expression.

```python
[expression  for item in iterable]
[expression  for item in iterable  if condition]
```

Read it as: *"give me `expression` for each `item` in `iterable`, but only when `condition` is true."*

````python
# Basic - squares of 1 to 5
squares = [x ** 2 for x in range(1, 6)]
print(squares)   # [1, 4, 9, 16, 25]

# Equivalent for loop:
squares2 = []
for x in range(1, 6):
    squares2.append(x ** 2)
print(squares2)
````

````python
# With a filter condition - even numbers only
evens = [x for x in range(1, 11) if x % 2 == 0]
print(evens)   # [2, 4, 6, 8, 10]
````

````python
# Transform a list of strings
words     = ["hello", "world", "python"]
upper     = [w.upper() for w in words]
lengths   = [len(w) for w in words]
long_ones = [w for w in words if len(w) > 4]

print(upper)       # ['HELLO', 'WORLD', 'PYTHON']
print(lengths)     # [5, 5, 6]
print(long_ones)   # ['hello', 'world', 'python']
````

````python
# if-else inside a comprehension - ternary goes BEFORE the for
numbers  = [1, -2, 3, -4, 5]
abs_vals = [x if x >= 0 else -x for x in numbers]
print(abs_vals)   # [1, 2, 3, 4, 5]
````

````python
# Nested comprehension - flatten a 2D list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat   = [elem for row in matrix for elem in row]
print(flat)   # [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Read the nested comprehension left to right:
# outer loop: for row in matrix
# inner loop: for elem in row
# collect:    elem
````

## 6. Nested Lists

A list whose elements are themselves lists - commonly used to represent **2D grids / matrices**.

Access: `matrix[row][col]` - first pick the row, then pick the column inside that row.

````python
# 3×3 matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matrix[0])       # [1, 2, 3]  - first row
print(matrix[1][2])    # 6          - row 1, col 2
print(matrix[-1][-1])  # 9          - bottom-right corner
````

````python
# Iterate over a matrix row by row
for row in matrix:
    for elem in row:
        print(elem, end=" ")
    print()
````

````python
# Create a matrix using comprehension (correct way)
rows, cols = 3, 4
grid = [[0] * cols for _ in range(rows)]   # 3×4 grid of zeros
print(grid)

# Common mistake - do NOT use multiplication for nested lists
wrong = [[0] * cols] * rows   # all rows are the SAME list object!
wrong[0][0] = 99
print("wrong:", wrong)        # every row changed!
````

````python
# Useful matrix operations
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

# Row sums
row_sums = [sum(row) for row in matrix]
print("Row sums:", row_sums)   # [6, 15, 24]

# Column sums (zip(*matrix) unpacks rows into column-wise tuples)
col_sums = [sum(col) for col in zip(*matrix)]
print("Col sums:", col_sums)   # [12, 15, 18]

# Transpose - rows become columns
transpose = [list(row) for row in zip(*matrix)]
for row in transpose:
    print(row)
````

````python
# Diagonal elements
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
n = len(matrix)

main_diag = [matrix[i][i]       for i in range(n)]   # top-left to bottom-right
anti_diag = [matrix[i][n-1-i]   for i in range(n)]   # top-right to bottom-left

print("Main diagonal:", main_diag)   # [1, 5, 9]
print("Anti diagonal:", anti_diag)   # [3, 5, 7]
````

## 7. Useful Built-in Functions

| Function | Returns | Example |
|----------|---------|----------|
| `len(lst)` | Number of elements | `len([1,2,3])` → `3` |
| `sum(lst)` | Sum of all elements | `sum([1,2,3])` → `6` |
| `min(lst)` | Smallest element | `min([3,1,2])` → `1` |
| `max(lst)` | Largest element | `max([3,1,2])` → `3` |
| `sorted(lst)` | New sorted list | `sorted([3,1,2])` → `[1,2,3]` |
| `reversed(lst)` | Reverse iterator | `list(reversed([1,2,3]))` → `[3,2,1]` |
| `any(lst)` | `True` if any element is truthy | `any([0,1,0])` → `True` |
| `all(lst)` | `True` if all elements are truthy | `all([1,2,0])` → `False` |

````python
nums = [4, 7, 2, 9, 1, 5]

print(len(nums))             # 6
print(sum(nums))             # 28
print(min(nums))             # 1
print(max(nums))             # 9
print(sorted(nums))          # [1, 2, 4, 5, 7, 9]  - new list, original unchanged
print(nums)                  # [4, 7, 2, 9, 1, 5]  - original intact
print(list(reversed(nums)))  # [5, 1, 9, 2, 7, 4]
````

````python
# sorted() with a key function
words = ["banana", "apple", "fig", "cherry", "date"]
print(sorted(words))                        # alphabetical
print(sorted(words, key=len))               # by length
print(sorted(words, key=len, reverse=True)) # by length, descending
````

````python
# any() and all()
nums = [2, 4, 6, 8]
print(all(n % 2 == 0 for n in nums))   # True - all even
print(any(n > 7 for n in nums))        # True - at least one > 7

mixed = [0, 1, 2, 3]
print(any(mixed))                       # True - 1, 2, 3 are truthy
print(all(mixed))                       # False - 0 is falsy
````

## 8. Applied Examples

### Example 1 - Remove duplicates while preserving order

````python
lst  = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3]
seen = []
for x in lst:
    if x not in seen:
        seen.append(x)
print(seen)   # [3, 1, 4, 5, 9, 2, 6]

# Pythonic - dict.fromkeys preserves insertion order (Python 3.7+)
print(list(dict.fromkeys(lst)))
````

### Example 2 - Rotate a list by k positions

````python
def rotate_left(lst, k):
    k = k % len(lst)        # handle k > len gracefully
    return lst[k:] + lst[:k]

nums = [1, 2, 3, 4, 5]
print(rotate_left(nums, 2))   # [3, 4, 5, 1, 2]
````

### Example 3 - Find the second largest element

````python
lst = [3, 1, 4, 1, 5, 9, 2, 6]

unique_sorted = sorted(set(lst), reverse=True)
print("Second largest:", unique_sorted[1])   # 6
````

### Example 4 - Check rows of a matrix against a target sum

````python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
target = 15

matching = [i for i, row in enumerate(matrix) if sum(row) == target]
print(f"Rows summing to {target}: {matching}")   # [1]
````

### Example 5 - Merge two sorted lists

````python
a = [1, 3, 5, 7]
b = [2, 4, 6, 8]

merged = sorted(a + b)
print(merged)   # [1, 2, 3, 4, 5, 6, 7, 8]
````

## Summary

| Operation | Syntax |
|-----------|--------|
| Create | `[]`, `list(range(n))`, `list("str")`, `[x]*n` |
| Index | `lst[i]`, `lst[-1]` |
| Slice | `lst[a:b]`, `lst[::step]`, `lst[::-1]` |
| Add | `append(x)`, `insert(i,x)`, `extend(iter)` |
| Remove | `remove(x)`, `pop(i)`, `del lst[i]` |
| Search | `index(x)`, `count(x)`, `x in lst` |
| Sort | `sort()` in-place · `sorted()` new list |
| Comprehension | `[expr for x in iter if cond]` |
| Nested access | `matrix[row][col]` |
| Aggregate | `sum()`, `min()`, `max()`, `any()`, `all()` |
| Transpose | `[list(r) for r in zip(*matrix)]` |

> **Key distinction:** `sort()` and `reverse()` modify the list **in place** and return `None`. `sorted()` and `list(reversed(...))` return a **new object** without touching the original.
