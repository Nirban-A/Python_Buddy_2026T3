## Question 1 - Rotate a List

```python
def rotate_right(lst: list, k: int) -> list:
    if not lst:
        return []
    k = k % len(lst)   # handle k > len
    return lst[-k:] + lst[:-k] if k else lst[:]

print(rotate_right([1, 2, 3, 4, 5], 2))  # [4, 5, 1, 2, 3]
print(rotate_right([1, 2, 3], 6))         # [1, 2, 3]
```

**Explanation:**

- `k % len(lst)` normalises the rotation so `k=6` on a 3-element list is the same as `k=0`.
- `lst[-k:]` takes the last `k` elements; `lst[:-k]` takes everything before them.
- Concatenating them puts the tail at the front - a right rotation.
- When `k=0` after normalisation, `lst[:-0]` would return `[]`, so we handle that with `lst[:]` instead.

```python
# Alternative using deque
from collections import deque

def rotate_right_v2(lst: list, k: int) -> list:
    d = deque(lst)
    d.rotate(k)
    return list(d)

print(rotate_right_v2([1, 2, 3, 4, 5], 2))
```

## Question 2 - Filter and Transform a List of Tuples

```python
def top_scorers(records: list, threshold: int) -> list:
    filtered = [(name, score) for name, score, _ in records if score >= threshold]
    return sorted(filtered, key=lambda x: (-x[1], x[0]))

records = [
    ("Alice", 85, "Math"),
    ("Bob",   72, "Science"),
    ("Carol", 91, "Math"),
    ("Dave",  85, "English"),
    ("Eve",   60, "Science"),
]
print(top_scorers(records, 80))
```

**Explanation:**

- The list comprehension filters and unpacks - `_` discards the subject since we don't need it.
- `sorted(key=lambda x: (-x[1], x[0]))` sorts by:
  - `-x[1]` (negated score) → highest score first
  - `x[0]` (name) → alphabetical on tie
- Tuple comparison is lexicographic - the two-element key handles both criteria in one `sorted()` call.

## Question 3 - Flatten and Filter Nested List

```python
def flatten_and_filter(matrix: list, threshold: int) -> list:
    return sorted(x for row in matrix for x in row if x > threshold)

matrix = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
print(flatten_and_filter(matrix, 4))   # [5, 5, 6, 9]
print(flatten_and_filter(matrix, 0))   # [1, 1, 2, 3, 4, 5, 5, 6, 9]
```

**Explanation:**

The nested generator `x for row in matrix for x in row` flattens the 2D list into a stream of individual numbers. The `if x > threshold` condition filters inline. `sorted()` sorts the results.

This is one of the most compact and readable ways to flatten + filter in Python.

```python
# Alternative - explicit loops
def flatten_and_filter_v2(matrix: list, threshold: int) -> list:
    result = []
    for row in matrix:
        for x in row:
            if x > threshold:
                result.append(x)
    return sorted(result)

print(flatten_and_filter_v2(matrix, 4))
```

## Question 4 - Divide Into Almost Equal Parts

```python
def divide_into_almost_equal_parts(n: int, k: int) -> list:
    base, remainder = divmod(n, k)
    return [base + 1] * remainder + [base] * (k - remainder)

print(divide_into_almost_equal_parts(5, 3))   # [2, 2, 1]
print(divide_into_almost_equal_parts(16, 3))  # [6, 5, 5]
print(divide_into_almost_equal_parts(12, 4))  # [3, 3, 3, 3]
```

**Explanation:**

- `divmod(n, k)` gives `(base, remainder)` - e.g. `divmod(5, 3)` = `(1, 2)`.
- Every part gets at least `base`. The leftover `remainder` units are distributed one each to the first `remainder` parts, making them `base + 1`.
- `[base+1] * remainder` creates the larger parts; `[base] * (k-remainder)` creates the smaller ones.
- Concatenating them puts larger values first - exactly what the problem requires.

**Verify:** `5, k=3` → `base=1, remainder=2` → `[2, 2] + [1]` → `[2, 2, 1]`. Sum = 5 ✓
