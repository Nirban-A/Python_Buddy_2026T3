## Question 1 — Word Frequency

```python
def word_frequency(words: list) -> dict:
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

# Test
print(word_frequency(["apple", "banana", "apple", "cherry", "banana", "apple"]))
print(word_frequency([]))
```

**Explanation:**

`freq.get(word, 0)` safely returns 0 if the word hasn't been seen yet. Adding 1 each time a word is encountered builds the frequency count.

**Alternative using Counter:**

```python
from collections import Counter

def word_frequency_v2(words: list) -> dict:
    return dict(Counter(words))

print(word_frequency_v2(["apple", "banana", "apple", "cherry", "banana", "apple"]))
```

## Question 2 — Apply Settings (Dictionary Unpacking)

```python
def apply_settings(defaults: dict, overrides: dict) -> dict:
    return {**defaults, **overrides}

# Test
d = {"theme": "light", "font": "Arial", "size": 14}
o = {"theme": "dark", "size": 16}
print(apply_settings(d, o))
```

**Explanation:**

`{**defaults, **overrides}` unpacks both dictionaries into a new one. When the same key appears in both, the later one (`overrides`) wins — exactly the "override" behaviour we want. The originals are never modified since a new dict is created.

## Question 3 — Top Subject Per Student (Nested Dictionary)

```python
def top_subject_per_student(data: dict) -> dict:
    result = {}
    for student, subjects in data.items():
        # max with tuple key: (-score, subject) → highest score first, alphabetical on tie
        top = min(subjects, key=lambda s: (-subjects[s], s))
        result[student] = top
    return result

# Test
data = {
    "Alice": {"Math": 85, "Science": 90, "English": 78},
    "Bob":   {"Math": 70, "Science": 70, "English": 85},
    "Carol": {"Math": 95, "Science": 80, "English": 95},
}
print(top_subject_per_student(data))
```

**Explanation:**

For each student, we need the subject with the max score; ties go to the alphabetically earlier subject.

`min(subjects, key=lambda s: (-subjects[s], s))` works because:
- `-subjects[s]` negates the score — `min` of negated scores = `max` of actual scores
- `s` (the subject name) is the tiebreaker — `min` of strings = alphabetically first

This is the "negate + min" trick for getting max with alphabetical tie-breaking in one call.

```python
# Alternative — using max with sorted for tie-breaking
def top_subject_per_student_v2(data: dict) -> dict:
    result = {}
    for student, subjects in data.items():
        max_score = max(subjects.values())
        # among all subjects with max score, pick the alphabetically first
        top = min(s for s in subjects if subjects[s] == max_score)
        result[student] = top
    return result

print(top_subject_per_student_v2(data))
```

## Question 4 — Group and Sum by Key

```python
def group_and_sum(entries: list) -> dict:
    totals = {}
    for entry in entries:
        key, vals_str = entry.split(":")
        vals = list(map(int, vals_str.split(",")))
        totals[key] = totals.get(key, 0) + sum(vals)
    return dict(sorted(totals.items()))

# Test
print(group_and_sum(["p:2,4", "q:1,3,5", "p:6", "r:0"]))
print(group_and_sum(["a:10", "b:5,5", "a:20,5"]))
```

**Explanation:**

1. `entry.split(":")` separates the key from the values string.
2. `map(int, vals_str.split(","))` converts the comma-separated values to integers.
3. `totals.get(key, 0) + sum(vals)` accumulates totals safely — starts at 0 for new keys.
4. `dict(sorted(totals.items()))` returns the result with keys in alphabetical order.
