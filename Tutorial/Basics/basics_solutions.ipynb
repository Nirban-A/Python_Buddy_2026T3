## Q1 · Classify a Number

**Core idea:** three-way if/elif/else on sign comparison.

```python
# --- Explanatory ---
def classify_number(n: int) -> str:
    if n > 0:
        return "positive"
    elif n < 0:
        return "negative"
    else:
        return "zero"

# --- Pythonic (ternary chain) ---
def classify_number_pythonic(n: int) -> str:
    return "positive" if n > 0 else ("negative" if n < 0 else "zero")

# --- Test ---
assert classify_number(5)    == "positive"
assert classify_number(-3)   == "negative"
assert classify_number(0)    == "zero"
print("Q1 passed ✓")
```

**Key takeaway:** Check `n > 0` first, then `n < 0`, and let `else` catch zero. The ternary chain is compact but the if/elif/else version is clearer for beginners.

## Q2 · Sum of Digits

**Core idea:** extract digits one by one using `% 10` and `// 10`, or convert to string and sum.

```python
# --- Explanatory (arithmetic) ---
def sum_of_digits(n: int) -> int:
    total = 0
    while n > 0:
        total += n % 10   # last digit
        n //= 10          # remove last digit
    return total

# --- Pythonic (string conversion) ---
def sum_of_digits_pythonic(n: int) -> int:
    return sum(int(d) for d in str(n))

# --- Test ---
assert sum_of_digits(123)  == 6
assert sum_of_digits(9999) == 36
assert sum_of_digits(0)    == 0
print("Q2 passed ✓")
```

**Key takeaway:** `n % 10` gives the last digit; `n //= 10` removes it. The while loop runs until all digits are consumed. The string approach `sum(int(d) for d in str(n))` is cleaner but the arithmetic version shows the underlying mechanism.

## Q3 · Count Vowels in a String

**Core idea:** iterate over each character and check membership in a vowel set.

```python
# --- Explanatory ---
def count_vowels(s: str) -> int:
    vowels = "aeiou"
    count  = 0
    for ch in s:
        if ch.lower() in vowels:
            count += 1
    return count

# --- Pythonic (sum + generator) ---
def count_vowels_pythonic(s: str) -> int:
    return sum(1 for ch in s if ch.lower() in "aeiou")

# --- Using lambda ---
count_vowels_lambda = lambda s: sum(1 for ch in s if ch.lower() in "aeiou")

# --- Test ---
assert count_vowels("Hello")  == 2
assert count_vowels("AEIOU")  == 5
assert count_vowels("xyz")    == 0
print("Q3 passed ✓")
```

**Key takeaway:** `ch.lower() in "aeiou"` handles case-insensitivity cleanly - lowercase once before checking so the vowel string only needs lowercase letters.

## Q4 · Multiplication Table

**Core idea:** loop from 1 to 10, build each string with an f-string.

```python
# --- Explanatory ---
def multiplication_table(n: int) -> list:
    result = []
    for i in range(1, 11):
        result.append(f"{n} x {i} = {n * i}")
    return result

# --- Pythonic (list comprehension) ---
def multiplication_table_pythonic(n: int) -> list:
    return [f"{n} x {i} = {n * i}" for i in range(1, 11)]

# --- Test ---
t = multiplication_table(3)
assert t[0]  == "3 x 1 = 3"
assert t[9]  == "3 x 10 = 30"
assert len(t) == 10
print("Q4 passed ✓")
```

**Key takeaway:** `range(1, 11)` produces 1 through 10 inclusive. f-strings make the formatting clean - `f"{n} x {i} = {n * i}"` reads exactly like the output.

## Q5 · Check Armstrong Number

**Core idea:** count digits, sum each digit raised to that power, compare with original.

```python
# --- Explanatory ---
def is_armstrong(n: int) -> bool:
    digits = str(n)
    power  = len(digits)
    total  = 0
    for d in digits:
        total += int(d) ** power
    return total == n

# --- Pythonic ---
def is_armstrong_pythonic(n: int) -> bool:
    s = str(n)
    return sum(int(d) ** len(s) for d in s) == n

# --- Using lambda ---
is_armstrong_lambda = lambda n: sum(int(d)**len(str(n)) for d in str(n)) == n

# --- Test ---
assert is_armstrong(153) == True
assert is_armstrong(370) == True
assert is_armstrong(100) == False
assert is_armstrong(9)   == True
print("Q5 passed ✓")
```

**Key takeaway:** `len(str(n))` gives the number of digits in one step. Raising each digit to that power and summing gives the Armstrong test. Converting to string first makes digit extraction simple.

## Q6 · Count Consecutive Character Groups

**Core idea:** compare each character to the previous one - start a new group whenever they differ.

```python
# --- Explanatory ---
def count_groups(s: str) -> int:
    if not s:
        return 0
    groups = 1
    for i in range(1, len(s)):
        if s[i] != s[i - 1]:
            groups += 1
    return groups

# --- Pythonic (zip adjacent pairs) ---
def count_groups_pythonic(s: str) -> int:
    if not s:
        return 0
    return 1 + sum(1 for a, b in zip(s, s[1:]) if a != b)

# --- Test ---
assert count_groups("aabbbcc") == 3
assert count_groups("abcd")    == 4
assert count_groups("aaaa")    == 1
assert count_groups("")        == 0
print("Q6 passed ✓")
```

**Key takeaway:** Start the count at 1 (first group always exists for non-empty string), then add 1 every time a character differs from the previous. `zip(s, s[1:])` is the clean pattern for checking adjacent pairs.

## Q7 · Collatz Sequence Length

**Core idea:** simulate the sequence with a while loop, count steps until n == 1.

```python
# --- Explanatory ---
def collatz_length(n: int) -> int:
    steps = 0
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        steps += 1
    return steps

# --- Pythonic (same logic, ternary) ---
def collatz_length_pythonic(n: int) -> int:
    steps = 0
    while n != 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps

# --- Test ---
assert collatz_length(1)  == 0
assert collatz_length(6)  == 8
assert collatz_length(27) == 111
print("Q7 passed ✓")
```

**Key takeaway:** The while loop runs until `n == 1`. Increment the step counter after each transformation. The ternary `n // 2 if n % 2 == 0 else 3 * n + 1` captures both rules in one line.

## Q8 · Caesar Cipher

**Core idea:** for each letter, shift its ordinal value by `shift` within the correct alphabet range using modulo.

```python
# --- Explanatory ---
def caesar_encrypt(text: str, shift: int) -> str:
    result = ""
    for ch in text:
        if ch.isalpha():
            base   = ord('A') if ch.isupper() else ord('a')
            # shift within 0-25, then add base
            result += chr((ord(ch) - base + shift) % 26 + base)
        else:
            result += ch
    return result

# --- Pythonic (join + generator) ---
def caesar_encrypt_pythonic(text: str, shift: int) -> str:
    def shift_char(ch):
        if not ch.isalpha():
            return ch
        base = ord('A') if ch.isupper() else ord('a')
        return chr((ord(ch) - base + shift) % 26 + base)
    return "".join(shift_char(ch) for ch in text)

# --- Test ---
assert caesar_encrypt("Hello, World!", 3) == "Khoor, Zruog!"
assert caesar_encrypt("xyz", 3)           == "abc"
assert caesar_encrypt("ABC", 2)           == "CDE"
assert caesar_encrypt("ZzAa", 1)          == "AaBb"
print("Q8 passed ✓")
```

**Key takeaway:** `(ord(ch) - base + shift) % 26` converts the character to 0-25, applies the shift with wrap-around, then `+ base` converts back to ASCII. Using separate bases for uppercase and lowercase preserves case.

## Q9 · Run-Length Encoding

**Core idea:** scan with a counter - when the character changes, write the previous character + count (omit count if 1).

```python
# --- Explanatory ---
def run_length_encode(s: str) -> str:
    if not s:
        return ""
    result  = ""
    current = s[0]
    count   = 1
    for ch in s[1:]:
        if ch == current:
            count += 1
        else:
            result += current + (str(count) if count > 1 else "")
            current = ch
            count   = 1
    # don't forget the last group
    result += current + (str(count) if count > 1 else "")
    return result

# --- Pythonic (using itertools.groupby) ---
from itertools import groupby

def run_length_encode_pythonic(s: str) -> str:
    result = ""
    for ch, group in groupby(s):
        count = sum(1 for _ in group)
        result += ch + (str(count) if count > 1 else "")
    return result

# --- Test ---
assert run_length_encode("aabbbcccc") == "a2b3c4"
assert run_length_encode("abcd")      == "abcd"
assert run_length_encode("aaabba")    == "a3b2a"
assert run_length_encode("")          == ""
print("Q9 passed ✓")
```

**Key takeaway:** Track `current` character and `count`. Flush the group whenever the character changes - and again after the loop ends for the final group. `itertools.groupby` handles the grouping automatically.

## Q10 · Number to Words (1–99)

**Core idea:** handle three cases - teens (11–19 have irregular names), tens (20, 30, …), and compound numbers (21–99 excluding teens).

```python
# --- Explanatory ---
def number_to_words(n: int) -> str:
    ones  = ["", "one", "two", "three", "four", "five",
              "six", "seven", "eight", "nine", "ten",
              "eleven", "twelve", "thirteen", "fourteen", "fifteen",
              "sixteen", "seventeen", "eighteen", "nineteen"]
    tens  = ["", "", "twenty", "thirty", "forty", "fifty",
              "sixty", "seventy", "eighty", "ninety"]

    if n < 20:
        return ones[n]
    elif n % 10 == 0:
        return tens[n // 10]
    else:
        return tens[n // 10] + " " + ones[n % 10]

# --- Test ---
assert number_to_words(5)  == "five"
assert number_to_words(13) == "thirteen"
assert number_to_words(20) == "twenty"
assert number_to_words(42) == "forty two"
assert number_to_words(99) == "ninety nine"
assert number_to_words(71) == "seventy one"
print("Q10 passed ✓")
```

**Key takeaway:** Index into two lookup lists - `ones[0..19]` for 1-19, `tens[2..9]` for multiples of ten. For compound numbers, join `tens[n//10]` and `ones[n%10]` with a space. The three-case if/elif/else maps directly to the structure of English number names.
