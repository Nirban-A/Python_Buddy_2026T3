## 1. Data Types

Python has several built-in data types. Every value has a type.

| Type | Example | Description |
|------|---------|-------------|
| `int` | `42`, `-7`, `0` | Whole numbers |
| `float` | `3.14`, `-0.5` | Decimal numbers |
| `str` | `"hello"` | Text |
| `bool` | `True`, `False` | Logical values |
| `NoneType` | `None` | Absence of a value |

````python
# int - whole numbers
age = 20
print(type(age), age)
````

````python
# float - decimal numbers
gpa = 8.75
print(type(gpa), gpa)
````

````python
# str - text
name = "Alice"
print(type(name), name, len(name))   # len() counts characters
````

````python
# bool - True or False
is_passed = True
print(type(is_passed), is_passed)
````

````python
# None - represents "no value"
result = None
print(type(result), result)
print(result is None)   # use 'is' to check for None
````

````python
# Type conversion
x = int("42")      # str → int
y = float("3.14")  # str → float
z = str(100)       # int → str
b = bool(0)        # 0, "", None, [] → False; everything else → True
print(x, y, z, b)
````

## 2. Variables and Expressions

````python
# Arithmetic operators
a, b = 17, 5
print(a + b)    # 22  - addition
print(a - b)    # 12  - subtraction
print(a * b)    # 85  - multiplication
print(a / b)    # 3.4 - true division (float)
print(a // b)   # 3   - floor division (integer)
print(a % b)    # 2   - modulo (remainder)
print(a ** 2)   # 289 - exponentiation
````

````python
# Comparison operators - return bool
print(5 > 3)    # True
print(5 == 5)   # True  (== checks equality)
print(5 != 4)   # True
print(5 <= 5)   # True
print(5 < 3)    # False
````

````python
# Logical operators
print(True and False)   # False - both must be True
print(True or False)    # True  - at least one must be True
print(not True)         # False

# Chained comparisons
x = 7
print(5 < x < 10)   # True - Pythonic range check
````

## 4. Conditionals - `if`, `elif`, `else`

```python
if condition1:
    ...
elif condition2:
    ...
else:
    ...
```

> Indentation defines the block - always use 4 spaces.

````python
# Basic if-else
score = 72
if score >= 60:
    print("Pass")
else:
    print("Fail")
````

````python
# if-elif-else chain
score = 85

if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "F"

print(f"Score: {score}, Grade: {grade}")
````

````python
# Nested if
num = -7
if num != 0:
    if num > 0:
        print("Positive")
    else:
        print("Negative")
else:
    print("Zero")
````

````python
# Ternary expression
age    = 17
status = "Adult" if age >= 18 else "Minor"
print(status)
````

````python
# FizzBuzz - classic OPPE pattern
# Always check combined case FIRST
n = 15
if n % 3 == 0 and n % 5 == 0:
    print("FizzBuzz")
elif n % 3 == 0:
    print("Fizz")
elif n % 5 == 0:
    print("Buzz")
else:
    print(n)
````

## 5. `while` Loops

```python
while condition:
    # body - must eventually make condition False
```

````python
# Count from 1 to 5
i = 1
while i <= 5:
    print(i, end=" ")
    i += 1
print()
````

````python
# Digit sum using while
n     = 1234
total = 0
while n > 0:
    total += n % 10   # last digit
    n //= 10          # remove last digit
print(f"Digit sum = {total}")   # 10
````

````python
# Reverse a number
num     = 12345
rev     = 0
original = num
while num > 0:
    rev = rev * 10 + num % 10
    num //= 10
print(f"Reverse of {original} = {rev}")
````

````python
# Count digits
n      = 98765
digits = 0
temp   = n
while temp > 0:
    digits += 1
    temp //= 10
print(f"{n} has {digits} digits")
````

## 6. `for` Loops

```python
for variable in sequence:
    # body - runs once per item
```

````python
# range() - the most common for-loop tool
# range(stop)              → 0 to stop-1
# range(start, stop)       → start to stop-1
# range(start, stop, step) → with custom step

for i in range(5):          print(i, end=" ")  # 0 1 2 3 4
print()
for i in range(1, 6):       print(i, end=" ")  # 1 2 3 4 5
print()
for i in range(0, 11, 2):   print(i, end=" ")  # 0 2 4 6 8 10
print()
for i in range(5, 0, -1):   print(i, end=" ")  # 5 4 3 2 1
print()
````

````python
# for over a string
word = "Python"
for ch in word:
    print(ch, end="-")
print()

# Count vowels
vowels = sum(1 for ch in word if ch.lower() in "aeiou")
print(f"Vowels in '{word}': {vowels}")
````

````python
# Accumulation - sum, product, max
nums = [4, 7, 2, 9, 1, 5]

total   = 0
maximum = nums[0]

for n in nums:
    total += n
    if n > maximum:
        maximum = n

print(f"Sum={total}, Max={maximum}")
````

````python
# enumerate() - index AND value together
fruits = ["apple", "banana", "cherry"]
for i, fruit in enumerate(fruits):
    print(f"{i}: {fruit}")
````

````python
# zip() - iterate two sequences in parallel
names  = ["Alice", "Bob", "Carol"]
scores = [85, 72, 91]
for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

````python
# Build a result string character by character
s = "Hello World 123"
digits_only = ""
for ch in s:
    if ch.isdigit():
        digits_only += ch
print(digits_only)   # 123
````

## 7. Loop Control - `break` and `continue`

````python
# break - exit the loop immediately
for i in range(10):
    if i == 5:
        break
    print(i, end=" ")
print()   # 0 1 2 3 4
````

````python
# continue - skip to the next iteration
for i in range(10):
    if i % 2 == 0:
        continue
    print(i, end=" ")
print()   # 1 3 5 7 9
````

````python
# break in while - stop when condition met
n = 1
total = 0
while True:        # infinite loop - needs break to exit
    total += n
    if total >= 50:
        break
    n += 1
print(f"Sum reached {total} at n={n}")
````

## 8. Nested Loops

````python
# Multiplication table (3×3)
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i}×{j}={i*j}", end="  ")
    print()
````

````python
# Triangle of stars
n = 5
for i in range(1, n + 1):
    print("*" * i)
````

````python
# Number pattern
for i in range(1, 6):
    for j in range(1, i + 1):
        print(j, end=" ")
    print()
````

````python
# Find all pairs that sum to a target
nums   = [1, 4, 7, 3, 6, 2]
target = 8

for i in range(len(nums)):
    for j in range(i + 1, len(nums)):
        if nums[i] + nums[j] == target:
            print(f"({nums[i]}, {nums[j]})")
````

| Concept | Key syntax |
