## 1. What is a Dictionary?

A dictionary maps **keys** to **values**. Think of it like a real dictionary - you look up a word (key) to get its definition (value).

```
{key: value, key: value, ...}
```

**Key properties:**
- Keys must be **unique** and **immutable** (strings, numbers, tuples)
- Values can be anything - lists, other dicts, functions, etc.
- Dictionaries are **ordered** (insertion order preserved since Python 3.7)
- Lookup by key is O(1) - extremely fast

````python
# A simple dictionary
student = {
    "name":  "Alice",
    "score": 85,
    "grade": "B"
}

print(student)
print(type(student))
````

## 2. Creating Dictionaries

Three ways to create a dictionary:

````python
# Method 1 - curly brace literal
d1 = {"a": 1, "b": 2, "c": 3}
print("Literal:", d1)
````

````python
# Method 2 - dict() constructor with keyword arguments
d2 = dict(a=1, b=2, c=3)
print("dict():", d2)
````

````python
# Method 3 - from a list of (key, value) pairs
pairs = [("x", 10), ("y", 20), ("z", 30)]
d3 = dict(pairs)
print("From pairs:", d3)
````

````python
# Empty dictionary
empty = {}
also_empty = dict()
print(empty, also_empty)
````

````python
# Keys can be any immutable type
mixed_keys = {
    "string_key": "value1",
    42:            "value2",
    (1, 2):        "value3",   # tuple as key
}
print(mixed_keys)
````

We can use tupple as a key but not a list as tupple is immutable where as list is mutable.

## 3. Accessing and Modifying Values

````python
d = {"name": "Bob", "age": 25, "city": "Delhi"}

# Access by key
print(d["name"])   # "Bob"
print(d["age"])    # 25
````

````python
# KeyError if key doesn't exist
try:
    print(d["score"])
except KeyError as e:
    print(f"KeyError: {e}")
````

````python
# Safe access with .get() - returns None (or default) if key missing
print(d.get("score"))           # None
print(d.get("score", 0))        # 0 (default)
print(d.get("name", "Unknown")) # "Bob" (key exists)
````

````python
# Add or update a key
d["score"] = 90         # add new key
d["age"]   = 26         # update existing key
print(d)
````

````python
# Delete a key
del d["city"]
print(d)

# pop() - delete and return the value
age = d.pop("age")
print(f"Removed age: {age}")
print(d)
````

````python
# Check if a key exists
d = {"a": 1, "b": 2}
print("a" in d)    # True
print("c" in d)    # False
print("c" not in d) # True
````

## 4. Essential Dictionary Methods

| Method | Returns | Description |
|--------|---------|-------------|
| `d.keys()` | dict_keys | All keys |
| `d.values()` | dict_values | All values |
| `d.items()` | dict_items | All (key, value) pairs |
| `d.get(k, default)` | value | Safe lookup |
| `d.pop(k, default)` | value | Remove and return |
| `d.setdefault(k, default)` | value | Get or set if missing |
| `d.update(other)` | None | Merge another dict in |
| `d.copy()` | dict | Shallow copy |
| `d.clear()` | None | Remove all items |

````python
d = {"a": 1, "b": 2, "c": 3}

print(list(d.keys()))    # ['a', 'b', 'c']
print(list(d.values()))  # [1, 2, 3]
print(list(d.items()))   # [('a',1), ('b',2), ('c',3)]
````

````python
# setdefault - get value if key exists, else set it to default and return default
d = {"a": 1}
print(d.setdefault("a", 99))   # 1 (key exists, unchanged)
print(d.setdefault("b", 99))   # 99 (key didn't exist, now d["b"] = 99)
print(d)
````

````python
# update - merge another dict (overwrites existing keys)
d1 = {"a": 1, "b": 2}
d2 = {"b": 99, "c": 3}
d1.update(d2)
print(d1)   # {'a': 1, 'b': 99, 'c': 3}
````

````python
# Merge two dicts - Python 3.9+ shorthand
d1 = {"a": 1, "b": 2}
d2 = {"b": 99, "c": 3}
merged = d1 | d2        # new dict, d2 values win on conflict
print(merged)
````

## 5. Iterating Over Dictionaries

````python
scores = {"Alice": 85, "Bob": 72, "Carol": 91}

# Iterate over keys (default)
for name in scores:
    print(name)
````

````python
# Iterate over values
for score in scores.values():
    print(score)
````

````python
# Iterate over (key, value) pairs - most common in exams
for name, score in scores.items():
    print(f"{name}: {score}")
````

````python
# Sorted iteration
for name in sorted(scores):           # sort by key
    print(f"{name}: {scores[name]}")

print()

for name, score in sorted(scores.items(), key=lambda x: x[1], reverse=True):  # sort by value
    print(f"{name}: {score}")
````

## 6. Dictionary Comprehensions

Just like list comprehensions, but produce a dictionary:

```python
{key_expr: value_expr for item in iterable if condition}
```

````python
# Square numbers 1-5
squares = {n: n**2 for n in range(1, 6)}
print(squares)  # {1:1, 2:4, 3:9, 4:16, 5:25}
````

````python
# Filter: only passing scores (>= 60)
scores = {"Alice": 85, "Bob": 45, "Carol": 91, "Dave": 58}
passing = {name: score for name, score in scores.items() if score >= 60}
print(passing)
````

````python
# Transform values - convert scores to letter grades
def grade(score):
    if score >= 90: return "A"
    if score >= 75: return "B"
    if score >= 60: return "C"
    return "F"

grades = {name: grade(score) for name, score in scores.items()}
print(grades)
````

````python
# From two lists using zip
names  = ["Alice", "Bob", "Carol"]
values = [85, 72, 91]
d = {name: val for name, val in zip(names, values)}
print(d)
````

## 7. Nested Dictionaries

Dictionary values can themselves be dictionaries - common in OPPE problems.

````python
students = {
    "S1": {"name": "Alice", "scores": {"Math": 85, "Science": 90}},
    "S2": {"name": "Bob",   "scores": {"Math": 72, "Science": 65}},
}

# Access nested value
print(students["S1"]["name"])               # "Alice"
print(students["S1"]["scores"]["Math"])     # 85
````

````python
# Add a new nested entry
students["S3"] = {"name": "Carol", "scores": {"Math": 91, "Science": 88}}

# Iterate over nested structure
for sid, info in students.items():
    avg = sum(info["scores"].values()) / len(info["scores"])
    print(f"{info['name']} ({sid}): avg = {avg:.1f}")
````

## 8. Common Patterns

### Pattern 1 - Counting (frequency map)

````python
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]

# Manual count with dict.get()
freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1
print(freq)
````

````python
# Most frequent item
most_common = max(freq, key=freq.get)
print(f"Most common: {most_common} ({freq[most_common]} times)")
````

### Pattern 2 - Grouping (accumulate into lists)

````python
records = [
    ("Math", "Alice"), ("Science", "Bob"),
    ("Math", "Carol"), ("Science", "Alice"),
    ("Math", "Bob")
]

# Group students by subject
groups = {}
for subject, student in records:
    groups.setdefault(subject, []).append(student)

print(groups)
````

````python
# Alternatively - the classic if-key-in check
groups2 = {}
for subject, student in records:
    if subject not in groups2:
        groups2[subject] = []
    groups2[subject].append(student)

print(groups2)
````

### Pattern 3 - Inverting a dictionary

````python
# Swap keys and values
original = {"a": 1, "b": 2, "c": 3}
inverted = {v: k for k, v in original.items()}
print(inverted)   # {1: 'a', 2: 'b', 3: 'c'}
````

### Pattern 4 - Accumulate totals

````python
sales = [
    ("Jan", 100), ("Feb", 150), ("Jan", 200),
    ("Mar", 120), ("Feb", 80),
]

totals = {}
for month, amount in sales:
    totals[month] = totals.get(month, 0) + amount

print(totals)  # {'Jan': 300, 'Feb': 230, 'Mar': 120}
````

## 9. `defaultdict` and `Counter`

The `collections` module provides specialised dict subclasses that handle common patterns more cleanly.

````python
from collections import defaultdict

# defaultdict - auto-initialises missing keys
# defaultdict(int)  → missing keys start at 0
# defaultdict(list) → missing keys start as []
# defaultdict(set)  → missing keys start as set()

freq = defaultdict(int)
for word in ["apple", "banana", "apple", "cherry"]:
    freq[word] += 1     # no .get() needed!

print(dict(freq))
````

````python
# defaultdict(list) - grouping made easy
groups = defaultdict(list)
for subject, student in [("Math","Alice"),("Science","Bob"),("Math","Carol")]:
    groups[subject].append(student)   # no setdefault needed!

print(dict(groups))
````

````python
from collections import Counter

# Counter - counts elements automatically
words = ["apple", "banana", "apple", "cherry", "banana", "apple"]
c = Counter(words)
print(c)

# Most common elements
print(c.most_common(2))   # [("apple", 3), ("banana", 2)]
````

````python
# Counter works on strings too
char_count = Counter("mississippi")
print(char_count)
print(char_count["s"])   # 4
````

````python
# Counter arithmetic
c1 = Counter({"a": 3, "b": 2})
c2 = Counter({"a": 1, "b": 5, "c": 2})

print(c1 + c2)   # add counts
print(c1 - c2)   # subtract (only positive results kept)
````

## 10. Applied Example - OPPE Question Walkthrough

### 2026 Jan Set 2 - S2Q2: Group and Sum Values by Key

**Problem:** Read `n` lines, each with a key and comma-separated values. Group by key, sum all values, print totals in sorted order.

```
Input:          Output:
4               p 12
p:2,4           q 9
q:1,3,5         r 0
p:6
r:0
```

This problem uses three dict patterns: **accumulation**, **sorted iteration**, and **safe default with .get()**.

````python
# ── Simulate input ─────────────────────────────────────────
input_lines = [
    "p:2,4",
    "q:1,3,5",
    "p:6",
    "r:0",
]

# ── Solution using dict.get() ───────────────────────────────
totals = {}
for line in input_lines:
    key, vals_str = line.split(":")
    vals = list(map(int, vals_str.split(",")))
    totals[key] = totals.get(key, 0) + sum(vals)

for key in sorted(totals):
    print(key, totals[key])
````

````python
# ── Same solution using defaultdict ────────────────────────
from collections import defaultdict

totals = defaultdict(int)
for line in input_lines:
    key, vals_str = line.split(":")
    totals[key] += sum(map(int, vals_str.split(",")))

for key in sorted(totals):
    print(key, totals[key])
````

### Tracing step by step

| Line | Key | Values | Line sum | `totals` after |
|------|-----|--------|----------|----------------|
| `p:2,4` | `p` | [2,4] | 6 | `{p:6}` |
| `q:1,3,5` | `q` | [1,3,5] | 9 | `{p:6, q:9}` |
| `p:6` | `p` | [6] | 6 | `{p:12, q:9}` |
| `r:0` | `r` | [0] | 0 | `{p:12, q:9, r:0}` |

Sorted keys: `p, q, r` → output matches expected ✓

## Summary

| Concept | Key syntax |
|---------|------------|
| Create | `{k:v}` or `dict(k=v)` or `dict([(k,v)])` |
| Access | `d[key]` or `d.get(key, default)` |
| Add/Update | `d[key] = value` |
| Delete | `del d[key]` or `d.pop(key)` |
| Check key | `key in d` |
| Iterate | `for k in d` / `d.values()` / `d.items()` |
| Comprehension | `{k:v for k,v in iterable if cond}` |
| Merge | `d1.update(d2)` or `d1 \| d2` |
| Count | `d.get(k, 0) + 1` or `Counter` |
| Group | `d.setdefault(k, []).append(v)` or `defaultdict(list)` |
| Invert | `{v: k for k, v in d.items()}` |
| Most frequent | `max(d, key=d.get)` |
| Sorted output | `for k in sorted(d)` |

> **Golden rule:** Use `d.get(key, default)` whenever a key might be missing - it's safer than `d[key]` and cleaner than an `if key in d` check.
