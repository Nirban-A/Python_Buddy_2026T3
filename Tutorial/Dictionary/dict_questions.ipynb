```python
deep_merge(d1, d2)
# → {"a": 1, "b": {"x": 10, "y": 99, "z": 30}, "c": 42, "d": 7}
#         ↑ b merged recursively        ↑ c overridden (d2 value is int)
```

````python
def deep_merge(d1, d2):
    """
    Recursively merges d2 into d1. d2 values take precedence on scalar conflicts.

    Parameters
    ----------
    d1, d2 : dict  - may be nested to any depth

    Returns
    -------
    dict  - merged result; d1 and d2 are not modified
    """
    pass
````

````python
# Test Q15
d1 = {"a": 1, "b": {"x": 10, "y": 20}, "c": {"p": 5}}
d2 = {"b": {"y": 99, "z": 30}, "c": 42, "d": 7}
assert deep_merge(d1, d2) == {"a": 1, "b": {"x": 10, "y": 99, "z": 30}, "c": 42, "d": 7}

# Originals unchanged
assert d1 == {"a": 1, "b": {"x": 10, "y": 20}, "c": {"p": 5}}
assert d2 == {"b": {"y": 99, "z": 30}, "c": 42, "d": 7}

# Flat dicts - d2 wins on conflict
assert deep_merge({"a": 1, "b": 2}, {"b": 9, "c": 3}) == {"a": 1, "b": 9, "c": 3}

# Three levels deep
assert deep_merge(
    {"x": {"y": {"z": 1}}},
    {"x": {"y": {"z": 2, "w": 3}}}
) == {"x": {"y": {"z": 2, "w": 3}}}

# Empty dicts
assert deep_merge({}, {"a": 1}) == {"a": 1}
assert deep_merge({"a": 1}, {}) == {"a": 1}
print("Q15 passed")
````
