## 1. What is a Dictionary?

A dictionary stores data as **key: value** pairs.

- Each **key** is unique — like a label
- Each **value** is the data attached to that label
- You look up values by their key, not by position

```python
# A dictionary describing a student
student = {
    "name":    "Alice",
    "roll_no": "CS101",
    "score":   88,
    "passed":  True
}

print(student)
```

## 2. Creating and Accessing a Dictionary

```python
# Create a dictionary
capitals = {
    "India":  "New Delhi",
    "France": "Paris",
    "Japan":  "Tokyo"
}

# Access a value using its key
print(capitals["India"])   # New Delhi
print(capitals["Japan"])   # Tokyo
```

```python
# Add a new key-value pair
capitals["Germany"] = "Berlin"
print(capitals)
```

```python
# Update an existing value
capitals["France"] = "Paris (updated)"
print(capitals["France"])
```

```python
# Safe access with .get() — returns None if key not found
print(capitals.get("India"))      # New Delhi
print(capitals.get("Brazil"))     # None
print(capitals.get("Brazil", "Not found"))  # Not found
```

```python
# Check if a key exists
print("India" in capitals)    # True
print("Brazil" in capitals)   # False
```

## 3. Iterating Over Keys

Two equivalent ways to loop over the keys of a dictionary.

```python
marks = {"Math": 85, "Science": 90, "English": 78}

# Method 1 — plain for loop (iterates over keys by default)
for subject in marks:
    print(subject)
```

```python
# Method 2 — using dict.keys() explicitly
for subject in marks.keys():
    print(subject)
```

```python
# Both are equivalent — the plain for loop is more Pythonic
# Use .keys() when you want to make the intent very explicit

# You can also convert keys to a list
key_list = list(marks.keys())
print(key_list)   # ['Math', 'Science', 'English']
```

```python
# Access the value inside the loop using the key
for subject in marks:
    print(f"{subject}: {marks[subject]}")
```

```python
# Iterating over sorted keys
for subject in sorted(marks):
    print(f"{subject}: {marks[subject]}")
```

---
## 4. Iterating Over Values — `dict.values()`

```python
marks = {"Math": 85, "Science": 90, "English": 78}

# Loop over just the values
for score in marks.values():
    print(score)
```

```python
# Useful when you only need the values — e.g. sum or average
total   = sum(marks.values())
average = total / len(marks)

print(f"Total: {total}")
print(f"Average: {average:.2f}")
```

---
## 5. Iterating Over Key-Value Pairs — `dict.items()`

`dict.items()` gives you both the key and value together — the most common iteration pattern in OPPE problems.

```python
marks = {"Math": 85, "Science": 90, "English": 78}

# Each iteration gives a (key, value) tuple
for subject, score in marks.items():
    print(f"{subject} → {score}")
```

```python
# Practical example — classify scores
for subject, score in marks.items():
    grade = "Pass" if score >= 80 else "Fail"
    print(f"{subject}: {score} ({grade})")
```

## 6. Dictionary Unpacking — `**`

The `**` operator **unpacks** a dictionary — spreading its key-value pairs into a function call or into another dictionary.

```python
# Use case 1 — passing a dict as keyword arguments to a function
def greet(name, age, city):
    print(f"Hello, I'm {name}, {age} years old, from {city}.")

info = {"name": "Alice", "age": 22, "city": "Mumbai"}

# Without unpacking — manual
greet(name=info["name"], age=info["age"], city=info["city"])

# With unpacking — clean
greet(**info)
```

```python
# Use case 2 — merging two dictionaries
defaults = {"theme": "light", "language": "English", "font": "Arial"}
user_prefs = {"theme": "dark", "font": "Consolas"}

# User preferences override defaults
config = {**defaults, **user_prefs}
print(config)
# {'theme': 'dark', 'language': 'English', 'font': 'Consolas'}
```

```python
# Use case 3 — adding extra keys while copying
base = {"a": 1, "b": 2}
extended = {**base, "c": 3, "d": 4}
print(extended)   # {'a': 1, 'b': 2, 'c': 3, 'd': 4}
print(base)       # {'a': 1, 'b': 2}  ← original unchanged
```

```python
# Later keys override earlier ones when keys clash
d = {**{"x": 1, "y": 2}, **{"y": 99, "z": 3}}
print(d)   # {'x': 1, 'y': 99, 'z': 3}  ← y overwritten
```

## 7. Dictionary with a List as a Value

A common pattern in OPPE — each key maps to a **list** of values (e.g. a student's multiple scores, or a subject's list of students).

```python
# Each student has a list of scores across tests
score_book = {
    "Alice": [85, 90, 78],
    "Bob":   [72, 65, 80],
    "Carol": [91, 88, 95]
}

# Access the list
print(score_book["Alice"])        # [85, 90, 78]

# Access a specific score
print(score_book["Alice"][0])     # 85 (first test)
print(score_book["Carol"][-1])    # 95 (last test)
```

```python
# Add a new score to an existing student
score_book["Alice"].append(92)
print(score_book["Alice"])   # [85, 90, 78, 92]
```

```python
# Add a completely new student
score_book["David"] = [70, 75]
print(score_book)
```

```python
# Iterate and compute average per student
for student, scores in score_book.items():
    avg = sum(scores) / len(scores)
    print(f"{student}: scores = {scores}, avg = {avg:.1f}")
```

```python
# Building a list-value dict by accumulating
# Group students by first letter of their name
students = ["Alice", "Bob", "Anna", "Brian", "Carol"]

grouped = {}
for name in students:
    letter = name[0]
    if letter not in grouped:
        grouped[letter] = []
    grouped[letter].append(name)

print(grouped)
# {'A': ['Alice', 'Anna'], 'B': ['Bob', 'Brian'], 'C': ['Carol']}
```

```python
# Cleaner with setdefault()
grouped2 = {}
for name in students:
    grouped2.setdefault(name[0], []).append(name)

print(grouped2)
```

## 8. Nested Dictionaries

A nested dictionary is a dictionary where the **values are themselves dictionaries**.

```python
# Each student is a dict inside the outer dict
class_data = {
    "S1": {"name": "Alice", "age": 20, "score": 88},
    "S2": {"name": "Bob",   "age": 21, "score": 72},
    "S3": {"name": "Carol", "age": 20, "score": 95},
}

print(class_data)
```

```python
# Access the inner dictionary
print(class_data["S1"])           # {'name': 'Alice', 'age': 20, 'score': 88}

# Access a specific field inside the inner dict
print(class_data["S1"]["name"])   # Alice
print(class_data["S3"]["score"])  # 95
```

```python
# Update a nested value
class_data["S2"]["score"] = 80
print(class_data["S2"])
```

```python
# Add a new field to an existing nested dict
class_data["S1"]["grade"] = "A"
print(class_data["S1"])
```

```python
# Iterating over a nested dictionary
for roll, info in class_data.items():
    print(f"Roll {roll}: {info['name']}, Score: {info['score']}")
```

```python
# Finding the topper (highest score)
topper_roll = max(class_data, key=lambda roll: class_data[roll]["score"])
print(f"Topper: {class_data[topper_roll]['name']} with {class_data[topper_roll]['score']}")
```

```python
# Three levels deep — subject → year → score
records = {
    "Math":    {"2024": 85, "2025": 90},
    "Science": {"2024": 78, "2025": 82},
}

# Access
print(records["Math"]["2025"])      # 90

# Iterate all three levels
for subject, year_scores in records.items():
    for year, score in year_scores.items():
        print(f"{subject} {year}: {score}")
```

## Quick Reference

| Task | Code |
|------|------|
| Create | `d = {"a": 1, "b": 2}` |
| Access | `d["a"]` or `d.get("a", default)` |
| Add / Update | `d["c"] = 3` |
| Loop over keys | `for k in d:` or `for k in d.keys():` |
| Loop over values | `for v in d.values():` |
| Loop over both | `for k, v in d.items():` |
| Unpack into function | `func(**d)` |
| Merge two dicts | `{**d1, **d2}` |
| List as value | `d["k"] = [1, 2, 3]` |
| Append to list value | `d["k"].append(4)` |
| Nested access | `d["outer"]["inner"]` |
| Group into lists | `d.setdefault(key, []).append(val)` |
