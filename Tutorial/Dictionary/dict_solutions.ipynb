## Q1 - Frequency Count

```python
# --- Explanatory ---
def word_frequency_v1(words):
    freq = {}
    for word in words:
        if word in freq:
            freq[word] += 1
        else:
            freq[word] = 1
    return freq

# --- Pythonic (dict.get with default) ---
def word_frequency_v2(words):
    freq = {}
    for word in words:
        freq[word] = freq.get(word, 0) + 1
    return freq

# Tests
for fn in [word_frequency_v1, word_frequency_v2]:
    assert fn(["apple", "banana", "apple", "cherry", "banana", "apple"]) == {"apple": 3, "banana": 2, "cherry": 1}
    assert fn(["hi", "hi", "hello"]) == {"hi": 2, "hello": 1}
    assert fn([]) == {}
print("Q1 all passed")
```

**Key ideas:**
- `dict.get(key, default)` returns the current count or 0 if the key is missing - avoids an explicit `if/else`.
- Always initialise accumulators before the loop. Pattern: `d[k] = d.get(k, 0) + 1` is the most common one-liner for counting.

## Q2 - Invert a Dictionary

```python
# --- Explanatory ---
def invert_dict_v1(d):
    result = {}
    for key, val in d.items():
        if val not in result:
            result[val] = []
        result[val].append(key)
    # sort each list of keys
    for val in result:
        result[val].sort()
    return result

# --- Pythonic (setdefault) ---
def invert_dict_v2(d):
    result = {}
    for key, val in d.items():
        result.setdefault(val, []).append(key)
    return {val: sorted(keys) for val, keys in result.items()}

# Tests
for fn in [invert_dict_v1, invert_dict_v2]:
    assert fn({"a": 1, "b": 2, "c": 1}) == {1: ["a", "c"], 2: ["b"]}
    assert fn({"x": 10, "y": 20})        == {10: ["x"], 20: ["y"]}
    assert fn({})                         == {}
    assert fn({"p": 5, "q": 5, "r": 5}) == {5: ["p", "q", "r"]}
print("Q2 all passed")
```

**Key ideas:**
- `dict.setdefault(key, default)` inserts `default` if `key` is absent, then returns the value - avoids the `if key not in d` check.
- Sorting the lists at the end gives deterministic output regardless of insertion order.

## Q3 ★ - Swap Last Characters of Two Dict Values

```python
# --- Explanatory ---
def swap_last_chars_v1(d, k1, k2):
    v1, v2  = d[k1], d[k2]        # read both values first
    last1   = v1[-1]               # save last chars before overwriting
    last2   = v2[-1]
    d[k1]   = v1[:-1] + last2     # all but last of v1 + last of v2
    d[k2]   = v2[:-1] + last1

# --- Pythonic (one-liner swap using tuple unpacking) ---
def swap_last_chars_v2(d, k1, k2):
    v1, v2 = d[k1], d[k2]
    d[k1], d[k2] = v1[:-1] + v2[-1], v2[:-1] + v1[-1]

# Tests
for fn in [swap_last_chars_v1, swap_last_chars_v2]:
    d1 = {"first": "apple", "second": "mango", "third": "banana"}
    fn(d1, "first", "second")
    assert d1 == {"first": "applo", "second": "mange", "third": "banana"}

    d2 = {"key1": "hello", "key2": "world"}
    fn(d2, "key1", "key2")
    assert d2 == {"key1": "helld", "key2": "worlo"}
print("Q3 all passed")
```

**Key ideas:**
- `v[-1]` = last character; `v[:-1]` = everything except the last character.
- Save **both** values before modifying - if you write `d[k1]` first and then read `d[k1][-1]`, you get the wrong character.
- Tuple unpacking `a, b = expr1, expr2` evaluates both right-hand sides before any assignment - safe atomic swap.

## Q4 ★ - Add Average-Key Entry In-Place

```python
# --- Explanatory ---
def add_average_key_diff_v1(d, k1, k2):
    new_key   = round((k1 + k2) / 2)   # rounded average of the two keys
    new_value = abs(d[k1] - d[k2])     # absolute difference of their values
    d[new_key] = new_value

# --- Pythonic (same - this problem has a natural one-liner form) ---
def add_average_key_diff_v2(d, k1, k2):
    d[round((k1 + k2) / 2)] = abs(d[k1] - d[k2])

# Tests
for fn in [add_average_key_diff_v1, add_average_key_diff_v2]:
    d1 = {2: 10, 8: 30}
    fn(d1, 2, 8)
    assert d1 == {2: 10, 8: 30, 5: 20}

    d2 = {1: 100, 9: 40}
    fn(d2, 1, 9)
    assert d2 == {1: 100, 9: 40, 5: 60}
print("Q4 all passed")
```

**Key ideas:**
- `round()` returns an `int` when given a `.0` float from integer division - the key type stays `int`.
- Read both `d[k1]` and `d[k2]` before inserting - safe even if `new_key` would overwrite one of them.

## Q5 ★ - Merge Two Dicts, Sum on Conflicts

```python
# --- Explanatory ---
def merge_dictionaries_v1(d1, d2):
    result = dict(d1)                  # start with a copy of d1
    for key, value in d2.items():
        if key in result:
            result[key] += value       # key in both → sum
        else:
            result[key] = value        # key only in d2 → insert
    return result

# --- Pythonic (dict.get) ---
def merge_dictionaries_v2(d1, d2):
    result = dict(d1)
    for key, value in d2.items():
        result[key] = result.get(key, 0) + value
    return result

# Tests
for fn in [merge_dictionaries_v1, merge_dictionaries_v2]:
    assert fn({"a": 1, "b": 2}, {"b": 3, "c": 4}) == {"a": 1, "b": 5, "c": 4}
    assert fn({"x": 10}, {"x": 5, "y": 8})         == {"x": 15, "y": 8}
    assert fn({}, {"a": 1})                         == {"a": 1}
    d1, d2 = {"a": 1, "b": 2}, {"b": 3, "c": 4}
    fn(d1, d2)
    assert d1 == {"a": 1, "b": 2}   # unchanged
print("Q5 all passed")
```

**Key ideas:**
- `dict(d1)` creates a **shallow copy** - mutations to `result` don't affect `d1`.
- `result.get(key, 0) + value` is the universal pattern: handles both new keys (returns 0) and existing ones.

## Q6 ★ - Most Frequent Numbers

```python
# --- Explanatory ---
def most_frequent_v1(nums):
    # Step 1: build frequency dict
    freq = {}
    for n in nums:
        freq[n] = freq.get(n, 0) + 1
    # Step 2: find max frequency
    max_freq = max(freq.values())
    # Step 3: collect all numbers with that frequency
    result = []
    for n, count in freq.items():
        if count == max_freq:
            result.append(n)
    return sorted(result)

# --- Pythonic ---
def most_frequent_v2(nums):
    freq     = {}
    for n in nums:
        freq[n] = freq.get(n, 0) + 1
    max_freq = max(freq.values())
    return sorted(k for k, v in freq.items() if v == max_freq)

# Tests
for fn in [most_frequent_v1, most_frequent_v2]:
    assert fn([3, 5, 3, 2, 5, 8, 3, 8, 5, 0]) == [3, 5]
    assert fn([1, 2, 2, 3, 3, 3])               == [3]
    assert fn([7, 7, 8, 8])                     == [7, 8]
    assert fn([42])                             == [42]
print("Q6 all passed")
```

**Key ideas:**
- Three-step pattern: **build freq dict → find max → filter**. Clean and reusable.
- `max(freq.values())` finds the peak count; then filter `freq.items()` by that peak.
- `sorted()` at the end ensures a deterministic, ascending result.

## Q7 ★ - Group Numbers by Remainder

```python
# --- Explanatory ---
def group_by_remainder_v1(nums, k):
    groups = {}
    for n in nums:
        r = n % k
        if r not in groups:
            groups[r] = []
        groups[r].append(n)
    # sort values and return in sorted key order
    return {r: sorted(groups[r]) for r in sorted(groups)}

# --- Pythonic (setdefault) ---
def group_by_remainder_v2(nums, k):
    groups = {}
    for n in nums:
        groups.setdefault(n % k, []).append(n)
    return {r: sorted(groups[r]) for r in sorted(groups)}

# Tests
for fn in [group_by_remainder_v1, group_by_remainder_v2]:
    assert fn([10, 3, 7, 15, 2, 8, 21, 6], 3) == {0: [3, 6, 15, 21], 1: [7, 10], 2: [2, 8]}
    assert fn([1, 2, 3, 4], 2)                 == {0: [2, 4], 1: [1, 3]}
    assert fn([6, 12, 18], 6)                  == {0: [6, 12, 18]}
    assert fn([5], 3)                          == {2: [5]}
print("Q7 all passed")
```

**Key ideas:**
- `setdefault(key, [])` is the go-to for building a dict of lists - returns the existing list or inserts an empty one.
- Dict comprehension `{r: sorted(...) for r in sorted(groups)}` both sorts the values and orders the keys in one pass.

## Q8 ★ - Group and Sum Values by String Key

```python
# --- Explanatory ---
def group_and_sum_v1(entries):
    totals = {}
    for entry in entries:
        key, vals_str = entry.split(":")          # split on colon
        values        = [int(v) for v in vals_str.split(",")]  # parse ints
        totals[key]   = totals.get(key, 0) + sum(values)
    return dict(sorted(totals.items()))            # sort by key

# --- Pythonic ---
def group_and_sum_v2(entries):
    totals = {}
    for entry in entries:
        key, vals = entry.split(":")
        totals[key] = totals.get(key, 0) + sum(map(int, vals.split(",")))
    return dict(sorted(totals.items()))

# Tests
for fn in [group_and_sum_v1, group_and_sum_v2]:
    assert fn(["p:2,4", "q:1,3,5", "p:6", "r:0"]) == {"p": 12, "q": 9, "r": 0}
    assert fn(["a:1,2", "b:5", "a:3"])             == {"a": 6, "b": 5}
    assert fn(["z:10"])                             == {"z": 10}
print("Q8 all passed")
```

**Key ideas:**
- `str.split(":")` and `str.split(",")` are the two parsing steps; always split on the outer delimiter first.
- `map(int, iterable)` applies `int()` to every element - more efficient than a list comprehension for just type conversion.
- `dict(sorted(d.items()))` is the idiomatic way to get a dict with sorted keys.

## Q9 ★ - First Top Performer by Average

```python
# --- Explanatory ---
def top_performer_v1(records):
    scores = {}    # name → list of scores
    order  = []    # first-appearance order
    for name, score in records:
        if name not in scores:
            scores[name] = []
            order.append(name)
        scores[name].append(score)

    best      = order[0]
    best_avg  = sum(scores[best]) / len(scores[best])
    for name in order[1:]:
        avg = sum(scores[name]) / len(scores[name])
        if avg > best_avg:            # strictly greater - first stays on tie
            best, best_avg = name, avg
    return best

# --- Pythonic ---
def top_performer_v2(records):
    scores, order = {}, []
    for name, score in records:
        if name not in scores:
            scores[name] = []
            order.append(name)
        scores[name].append(score)
    # max() keeps the first occurrence when key values are equal
    return max(order, key=lambda n: sum(scores[n]) / len(scores[n]))

# Tests
for fn in [top_performer_v1, top_performer_v2]:
    assert fn([("Alice", 80), ("Bob", 90), ("Alice", 100), ("Bob", 70), ("Carol", 95)]) == "Carol"
    assert fn([("X", 50), ("Y", 50)])                                                   == "X"
    assert fn([("A", 100)])                                                             == "A"
    assert fn([("M", 60), ("N", 80), ("M", 100)])                                      == "M"
print("Q9 all passed")
```

**Key ideas:**
- Use a separate `order` list to track first-appearance; dicts alone don't guarantee insertion order for this logic.
- `max()` is **stable for first occurrence** - it only updates the winner on a **strictly greater** value, so ties keep the first.
- Compute average inside the key function - avoids storing a separate averages dict.

## Q10 - Merge List of Dicts, Keep Maximum on Conflicts

```python
# --- Explanatory ---
def merge_max_v1(dicts):
    result = {}
    for d in dicts:
        for key, value in d.items():
            if key not in result or value > result[key]:
                result[key] = value
    return result

# --- Pythonic ---
def merge_max_v2(dicts):
    result = {}
    for d in dicts:
        for key, value in d.items():
            result[key] = max(result.get(key, value), value)
    return result

# Tests
for fn in [merge_max_v1, merge_max_v2]:
    assert fn([{"a": 3, "b": 5}, {"b": 8, "c": 2}, {"a": 7, "c": 1}]) == {"a": 7, "b": 8, "c": 2}
    assert fn([{"x": 1}, {"x": 1}, {"x": 1}])                         == {"x": 1}
    assert fn([{}])                                                     == {}
    assert fn([{"p": 5}, {"q": 3}])                                    == {"p": 5, "q": 3}
print("Q10 all passed")
```

**Key ideas:**
- Iterate each dict in the list and update `result` only when the new value is larger.
- `result.get(key, value)` - using the new `value` as the default means a missing key always wins the `max()`, equivalent to just inserting it.

## Q11 - Top Subject Per Student

```python
# --- Explanatory ---
def top_subject_per_student_v1(data):
    result = {}
    for student, subjects in data.items():
        best_subject = None
        best_score   = -1
        for subject in sorted(subjects):       # sorted → alphabetical tiebreak
            if subjects[subject] > best_score:
                best_score   = subjects[subject]
                best_subject = subject
        result[student] = best_subject
    return result

# --- Pythonic ---
def top_subject_per_student_v2(data):
    return {
        student: max(sorted(subjects), key=lambda s: subjects[s])
        for student, subjects in data.items()
    }

# Tests
data1 = {
    "Alice": {"Math": 90, "Science": 85, "English": 90},
    "Bob":   {"Math": 70, "Science": 95, "English": 80},
}
for fn in [top_subject_per_student_v1, top_subject_per_student_v2]:
    assert fn(data1) == {"Alice": "English", "Bob": "Science"}
    data2 = {"X": {"A": 100}, "Y": {"B": 50, "C": 50}}
    assert fn(data2) == {"X": "A", "Y": "B"}
print("Q11 all passed")
```

**Key ideas:**
- `sorted(subjects)` gives subjects in alphabetical order; iterating in this order means the first one encountered at the max score wins - that's the alphabetical tiebreak.
- `max(sorted(subjects), key=lambda s: subjects[s])` - `max()` keeps the first maximum, and since subjects are sorted, the first is alphabetical.

## Q12 ★ - Process Orders with Threshold

```python
# --- Explanatory ---
def process_orders_v1(data, min_quantity):
    result = {}
    for record in data:
        customer = record["customer"]
        qtys     = [order["qty"] for order in record["orders"]]
        if not qtys:                  # skip customers with no orders
            continue
        total = sum(qtys)
        if total >= min_quantity:
            result[customer] = {
                "total":   total,
                "count":   len(qtys),
                "average": total / len(qtys),
                "max":     max(qtys),
            }
    return result

# --- Pythonic (same logic, slightly compressed) ---
def process_orders_v2(data, min_quantity):
    result = {}
    for rec in data:
        qtys = [o["qty"] for o in rec["orders"]]
        if not qtys:
            continue
        total = sum(qtys)
        if total >= min_quantity:
            n = len(qtys)
            result[rec["customer"]] = {"total": total, "count": n,
                                        "average": total / n, "max": max(qtys)}
    return result

# Tests
data1 = [
    {"customer": "c1", "orders": [{"qty": 2}, {"qty": 5}]},
    {"customer": "c2", "orders": [{"qty": 1}, {"qty": 1}]},
]
for fn in [process_orders_v1, process_orders_v2]:
    assert fn(data1, 5) == {"c1": {"total": 7, "count": 2, "average": 3.5, "max": 5}}

data2 = [
    {"customer": "A", "orders": [{"qty": 10}, {"qty": 10}]},
    {"customer": "B", "orders": []},
    {"customer": "C", "orders": [{"qty": 3}]},
]
for fn in [process_orders_v1, process_orders_v2]:
    r = fn(data2, 5)
    assert "A" in r and "B" not in r and "C" not in r
print("Q12 all passed")
```

**Key ideas:**
- Extract the inner `qty` values first with a list comprehension, then compute all aggregates from that list - avoids repeated traversal.
- `if not qtys: continue` guards against empty orders before calling `max()` or dividing.
- Filter (`if total >= min_quantity`) after computing - only build the result dict for qualifying customers.

## Q13 ★ - Railway Booking Analysis

```python
BERTH_TYPES = ["UB", "MB", "LB", "SU", "SL", None]

# --- Explanatory ---
def analyze_bookings_v1(bookings, task):
    if task == "ticket_prices":
        result = {}
        for pnr, b in bookings.items():
            result[pnr] = b["ticket_rate"] * len(b["passengers"])
        return result

    if task == "total_revenue":
        total = 0
        for b in bookings.values():
            total += b["ticket_rate"] * len(b["passengers"])
        return total

    if task == "berth_preference_count":
        counts = {bt: 0 for bt in BERTH_TYPES}   # all 6 types, start at 0
        for b in bookings.values():
            for _, pref in b["passengers"]:
                counts[pref] += 1
        return counts

    if task == "coach_type_count":
        counts = {}
        for b in bookings.values():
            ct = b["coach_type"]
            counts[ct] = counts.get(ct, 0) + len(b["passengers"])
        return counts

# --- Pythonic ---
def analyze_bookings_v2(bookings, task):
    if task == "ticket_prices":
        return {pnr: b["ticket_rate"] * len(b["passengers"]) for pnr, b in bookings.items()}

    if task == "total_revenue":
        return sum(b["ticket_rate"] * len(b["passengers"]) for b in bookings.values())

    if task == "berth_preference_count":
        counts = {bt: 0 for bt in BERTH_TYPES}
        for b in bookings.values():
            for _, pref in b["passengers"]:
                counts[pref] += 1
        return counts

    if task == "coach_type_count":
        counts = {}
        for b in bookings.values():
            ct = b["coach_type"]
            counts[ct] = counts.get(ct, 0) + len(b["passengers"])
        return counts

# Tests
bookings = {
    "PNR112": {"coach_type": "3AC", "ticket_rate": 900,
               "passengers": [("Rahul", "UB"), ("Baskar", None)]},
    "PNR123": {"coach_type": "2AC", "ticket_rate": 1500,
               "passengers": [("Amit", "LB"), ("Ravi", "UB")]},
    "PNR456": {"coach_type": "3AC", "ticket_rate": 800,
               "passengers": [("Priya", "MB"), ("Neha", None)]},
    "PNR789": {"coach_type": "SL",  "ticket_rate": 500,
               "passengers": [("Kiran", "SL"), ("Diya", "SU"), ("Arjun", "LB")]},
}
for fn in [analyze_bookings_v1, analyze_bookings_v2]:
    assert fn(bookings, "ticket_prices") == {"PNR112": 1800, "PNR123": 3000, "PNR456": 1600, "PNR789": 1500}
    assert fn(bookings, "total_revenue") == 7900
    berth = fn(bookings, "berth_preference_count")
    assert berth["UB"] == 2 and berth["LB"] == 2 and berth[None] == 2
    assert fn(bookings, "coach_type_count") == {"3AC": 4, "2AC": 2, "SL": 3}
print("Q13 all passed")
```

**Key ideas:**
- Pre-initialise `berth_preference_count` with all 6 types (including `None`) so absent types show 0 - critical for the spec.
- `None` is a valid dictionary key in Python.
- Separating tasks with `if task == "..":` blocks keeps each operation readable and independently testable.

## Q14 ★ - Sales Data Analysis

```python
# --- Explanatory ---
def analyse_sales_data_v1(sales_data, task):
    # Build one aggregation dict reused by all tasks
    # agg[pid] = [total_units, total_revenue]
    agg = {}
    for record in sales_data:
        pid = record["product_id"]
        if pid not in agg:
            agg[pid] = [0, 0]
        agg[pid][0] += record["units_sold"]
        agg[pid][1] += record["revenue"]

    if task == "total_revenue":
        return sum(v[1] for v in agg.values())

    if task == "product_wise_total_units_and_revenue":
        return {pid: (v[0], v[1]) for pid, v in agg.items()}

    if task == "top_selling_product":
        return max(agg, key=lambda pid: (agg[pid][0], agg[pid][1]))

    if task == "average_product_price":
        return {pid: round(v[1] / v[0], 2) for pid, v in agg.items()}

# --- Pythonic (identical structure; showing dict.get pattern for agg) ---
def analyse_sales_data_v2(sales_data, task):
    agg = {}
    for r in sales_data:
        pid = r["product_id"]
        u, rev = agg.get(pid, (0, 0))
        agg[pid] = (u + r["units_sold"], rev + r["revenue"])

    if task == "total_revenue":
        return sum(rev for _, rev in agg.values())
    if task == "product_wise_total_units_and_revenue":
        return dict(agg)
    if task == "top_selling_product":
        return max(agg, key=lambda pid: agg[pid])
    if task == "average_product_price":
        return {pid: round(rev / u, 2) for pid, (u, rev) in agg.items()}

# Tests
sales_data = [
    {"product_id": "P101", "units_sold": 50,  "revenue": 400},
    {"product_id": "P102", "units_sold": 30,  "revenue": 900},
    {"product_id": "P101", "units_sold": 70,  "revenue": 600},
    {"product_id": "P103", "units_sold": 120, "revenue": 600},
]
for fn in [analyse_sales_data_v1, analyse_sales_data_v2]:
    assert fn(sales_data, "total_revenue") == 2500
    assert fn(sales_data, "product_wise_total_units_and_revenue") == {"P101": (120, 1000), "P102": (30, 900), "P103": (120, 600)}
    assert fn(sales_data, "top_selling_product") == "P101"
    avg = fn(sales_data, "average_product_price")
    assert avg["P102"] == 30.0 and avg["P103"] == 5.0
print("Q14 all passed")
```

**Key ideas:**
- Build one aggregation dict once and reuse it across all four task branches - avoids scanning `sales_data` four times.
- Tuple comparison `(units, revenue)` in `max()` gives the tiebreak automatically: higher units first, then higher revenue.
- `round(rev / u, 2)` computes average price per unit (revenue ÷ units).

## Q15 - Deep-Merge Two Nested Dictionaries

```python
# --- Explanatory (recursive) ---
def deep_merge_v1(d1, d2):
    result = dict(d1)                       # start with a shallow copy of d1
    for key, val2 in d2.items():
        if key in result and isinstance(result[key], dict) and isinstance(val2, dict):
            # Both sides are dicts → recurse
            result[key] = deep_merge_v1(result[key], val2)
        else:
            # Key not in d1, or at least one side is not a dict → d2 wins
            result[key] = val2
    return result

# --- Pythonic (same recursion, slightly shorter) ---
def deep_merge_v2(d1, d2):
    result = {**d1}     # unpack d1 into result
    for k, v in d2.items():
        result[k] = (
            deep_merge_v2(result[k], v)
            if isinstance(result.get(k), dict) and isinstance(v, dict)
            else v
        )
    return result

# Tests
for fn in [deep_merge_v1, deep_merge_v2]:
    d1 = {"a": 1, "b": {"x": 10, "y": 20}, "c": {"p": 5}}
    d2 = {"b": {"y": 99, "z": 30}, "c": 42, "d": 7}
    assert fn(d1, d2) == {"a": 1, "b": {"x": 10, "y": 99, "z": 30}, "c": 42, "d": 7}
    # Originals unchanged
    assert d1 == {"a": 1, "b": {"x": 10, "y": 20}, "c": {"p": 5}}
    assert d2 == {"b": {"y": 99, "z": 30}, "c": 42, "d": 7}
    assert fn({"a": 1, "b": 2}, {"b": 9, "c": 3}) == {"a": 1, "b": 9, "c": 3}
    assert fn({"x": {"y": {"z": 1}}}, {"x": {"y": {"z": 2, "w": 3}}}) == {"x": {"y": {"z": 2, "w": 3}}}
    assert fn({}, {"a": 1}) == {"a": 1}
    assert fn({"a": 1}, {}) == {"a": 1}
print("Q15 all passed")
```

**Key ideas:**
- **Recursion** is the natural fit: each key either needs merging (recurse) or overriding (base case).
- `isinstance(v, dict)` checks type at runtime - both sides must be dicts to recurse; if either is a scalar, `d2` wins.
- `{**d1}` (dict unpacking) is a clean way to make a shallow copy without calling `dict(d1)`.
- The shallow copy is safe here because we only replace keys (never mutate sub-dicts of `d1` in-place).
